```python
def is_prime(n):
    if n <= 1:
        return False
    

    for i in range(2, n):
        if n % i == 0:
            return False  
            
    return True  
```

```python
print(is_prime(35))
```

```python
def print_fibonacci(n):
    if n <= 0:
        print("Please enter a positive integer.")
        return

    a, b = 0, 1

    print(f"Fibonacci sequence up to {n} terms:")

    for i in range(n):
        print(a, end=" ")
        a, b = b, a + b
    print() 


# Example usage:
terms = int(input("Enter the number of terms: "))
print_fibonacci(terms)
```

```python
def remove_duplicates(nums):
    seen = set()
    result = []
    for x in nums:
        if x not in seen:
            seen.add(x)
            result.append(x)
    return result
```

```python
remove_duplicates([1,1,2,3])
```

```python
def multiply(*args):
    res = 1
    for num in args:
        res *= num
    return res
```

```python
multiply(2, 3, 4)
```

```python
text = "Mubashir"
freq = {ch: text.count(ch) for ch in set(text)}
print(freq)
```
